# Exploratory Data Analysis (EDA)
In this notebook, we explore the Student Performance dataset to understand distributions and correlations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Ensure plots dir exists
os.makedirs('../outputs/plots', exist_ok=True)

# Load dataset
df = pd.read_csv('../data/student-mat.csv', sep=';')
df.head()
```

```python
# 1. Target Distribution
plt.figure(figsize=(8, 5))
sns.histplot(df['G3'], bins=20, kde=True)
plt.title('Distribution of Final Score (G3)')
plt.xlabel('Score')
plt.savefig('../outputs/plots/target_distribution.png')
plt.show()
```

```python
# 2. Correlation Matrix
plt.figure(figsize=(10, 8))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.savefig('../outputs/plots/correlation_matrix.png')
plt.show()
```

```python
# 3. Boxplot of Scores by Study Time
plt.figure(figsize=(8, 5))
sns.boxplot(x='studytime', y='G3', data=df)
plt.title('Final Score by Study Time')
plt.savefig('../outputs/plots/boxplot_studytime_G3.png')
plt.show()
```
